```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd 
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
df.drop(columns=['Status','unnamed1'],inplace=True) # this querry use of drop columns 
df.head()
```

# BAR CHART

# Total Sales Amount By Product Category

```python
sales_by_category = df.groupby('Product_Category')['Amount'].sum().reset_index()
sns.barplot(x='Amount', y='Product_Category', data=sales_by_category, palette='Blues_r')
plt.title('Total Sales Amount by Product Category')
plt.Figure(figsize=(12,8))
plt.xlabel('Total Amount')
plt.ylabel('Product Category')
plt.show()
```

# TOTAL ORDERS TOP (10) STATE

```python
top_states = df.groupby('State')['Orders'].sum().nlargest(10).reset_index()
sns.barplot(x='Orders', y='State', data=top_states, palette='viridis')
plt.title('Top 10 States by Total Orders')
plt.show()
```

# COUNT PLOT

```python
sns.countplot(x='Gender', data=df, palette='Set2')
plt.title('Distribution of Customers by Gender')
plt.show()
```

```python
sns.countplot(x='Age Group', data=df, order=sorted(df['Age Group'].unique()), palette='viridis')
plt.title('Customer Count Across Age Groups')
plt.show()
```

# PIE CHART

```python
zone_sales = df.groupby('Zone')['Amount'].sum()
plt.figure(figsize=(8,5))
plt.pie(zone_sales, labels=zone_sales.index, autopct='%1.1f%%', startangle=140, colors=sns.color_palette('pastel'))
plt.title('Percentage of Sales Amount by Zone')
plt.show()
```

# GROUPED STACK BAR CHART

```python
sns.barplot(x='Age Group', y='Amount', hue='Gender', data=df, estimator=sum, ci=None, palette='Blues_r')
plt.title('Total Sales Amount by Age Group and Gender')
plt.ylabel('Total Amount')
plt.show()
```

# BOX PLOT

```python
sns.barplot(x='Age Group', y='Amount', hue='Gender', data=df, estimator=sum, ci=None, palette='Set2')
plt.title('Total Sales Amount by Age Group and Gender')
plt.ylabel('Total Amount')
plt.show()
```

# BOX PLOT

```python
plt.figure(figsize=(12, 5))
sns.boxplot(x='Occupation', y='Amount', data=df, palette='Set3')
plt.xticks(rotation=45)
plt.title('Amount Spent Distribution Across Occupations')
plt.show()
```
